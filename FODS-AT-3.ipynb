```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("education.csv")

print("Rows and Columns:", df.shape)

print("\nData Types")
print(df.dtypes)

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Values")
print(df.duplicated().sum())

print("\nStatistical Summary")
print(df.describe())

df["Result"].value_counts().plot(kind="bar")
plt.title("Pass and Fail Count")
plt.show()

plt.hist(df["Attendance"], bins=5)
plt.title("Attendance Distribution")
plt.xlabel("Attendance")
plt.ylabel("Frequency")
plt.show()

plt.scatter(df["Study_Hours"], df["Internal_Marks"])
plt.xlabel("Study Hours")
plt.ylabel("Internal Marks")
plt.title("Study Hours vs Internal Marks")
plt.show()

sns.boxplot(y=df["Attendance"])
plt.title("Attendance Box Plot")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("retail_sales.csv")

print("Rows and Columns:", df.shape)

print("\nData Types")
print(df.dtypes)

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Values")
print(df.duplicated().sum())

print("\nStatistical Summary")
print(df.describe())

plt.figure(figsize=(7,5))
df.groupby("Category")["Revenue"].sum().plot(kind="bar")
plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.show()

plt.figure(figsize=(6,4))
plt.hist(df["Price"], bins=5)
plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df["Quantity_Sold"], df["Revenue"])
plt.title("Quantity Sold vs Revenue")
plt.xlabel("Quantity Sold")
plt.ylabel("Revenue")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(y=df["Discount"])
plt.title("Discount Box Plot")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("banking_loan.csv")

print("Rows and Columns:", df.shape)

print("\nData Types")
print(df.dtypes)

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Values")
print(df.duplicated().sum())

print("\nStatistical Summary")
print(df.describe())

plt.figure(figsize=(6,4))
df["Loan_Status"].value_counts().plot(kind="bar")
plt.title("Loan Approval Status")
plt.xlabel("Loan Status")
plt.ylabel("Count")
plt.show()

plt.figure(figsize=(6,4))
plt.hist(df["Credit_Score"], bins=5)
plt.title("Credit Score Distribution")
plt.xlabel("Credit Score")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df["Income"], df["Loan_Amount"])
plt.title("Income vs Loan Amount")
plt.xlabel("Income")
plt.ylabel("Loan Amount")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(y=df["Income"])
plt.title("Income Box Plot")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("agriculture_crop_yield.csv")

print("Rows and Columns:", df.shape)

print("\nData Types")
print(df.dtypes)

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Values")
print(df.duplicated().sum())

print("\nStatistical Summary")
print(df.describe())

plt.figure(figsize=(6,4))
df.groupby("Soil_Type")["Crop_Yield_kg"].mean().plot(kind="bar")
plt.title("Average Crop Yield by Soil Type")
plt.xlabel("Soil Type")
plt.ylabel("Crop Yield (kg)")
plt.show()

plt.figure(figsize=(6,4))
plt.hist(df["Rainfall_mm"], bins=5)
plt.title("Rainfall Distribution")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df["Rainfall_mm"], df["Crop_Yield_kg"])
plt.title("Rainfall vs Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Crop Yield (kg)")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(y=df["Fertilizer_kg"])
plt.title("Fertilizer Usage Box Plot")
plt.show()
```
